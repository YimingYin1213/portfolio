## Megalodon Boss Explanation
The megalodon boss fight uses the same CSSE topics as the rest of the Aquatic level, but in a more combat-focused way.
Instead of one long summary, this section breaks the boss code into smaller pieces so each topic can be seen directly in the code.

```python
this.startMegalodonEncounter = async () => {
  if (this.bossState.active || this.bossState.introPlayed) return;

  this.bossState.introPlayed = true;
  this.bossState.active = true;
  this.bossState.hp = this.bossState.maxHp;
  this.bossState.summonThresholdsTriggered = [];

  const bossData = {
    id: 'MegalodonBoss',
    src: this.bossState.megalodonMoveSheet,
    INIT_POSITION: { x: 120, y: this.gameEnv.innerHeight - 150 },
    laserAttack: { row: 0, start: 0, columns: 3 },
    rocketAttack: { row: 1, start: 0, columns: 3 }
  };

  const boss = new Npc(bossData, this.gameEnv);
  this.bossState.megalodon = boss;
  this.gameEnv.gameObjects.push(boss);
};

const thresholds = [0.75, 0.5, 0.25];
thresholds.forEach((threshold) => {
  if (this.bossState.hp <= this.bossState.maxHp * threshold) {
    summonRushingSharks();
  }
});
```

### Code Breakdown
`this.startMegalodonEncounter = async () => {`
This is a **function**. It starts the boss fight when the level reaches the megalodon scene.
`if (this.bossState.active || this.bossState.introPlayed) return;`
This is a **conditional** using **booleans**. It prevents the fight from starting twice.
`this.bossState.introPlayed = true;` and `this.bossState.active = true;`
These lines update **boolean** values so the game knows the intro already happened and the boss battle is now active.
`this.bossState.hp = this.bossState.maxHp;`
This uses **variables** and **data abstraction**. Both `hp` and `maxHp` are stored inside the `bossState` object.
`this.bossState.summonThresholdsTriggered = [];`
This is an **array**. The game uses it to remember which health checkpoints have already triggered extra shark summons.
`const bossData = { ... }`
This object is **data abstraction** because it groups the boss sprite, position, attacks, and identity into one structure.
`id: 'MegalodonBoss'` and `src: this.bossState.megalodonMoveSheet`
These are **strings** and **variables**. The string gives the boss a name, and the variable gives the boss its sprite sheet.
`INIT_POSITION: { x: 120, y: this.gameEnv.innerHeight - 150 }`
This uses **mathematical expressions**. The boss is placed by calculating a y-position from the game window height.
`const boss = new Npc(bossData, this.gameEnv);`
This shows **classes** and **constructors**. `Npc` is a class, and `new Npc(...)` calls its constructor to create the megalodon object.
`this.bossState.megalodon = boss;`
This stores the new boss in a variable inside `bossState` so other parts of the game can update or remove it later.
`this.gameEnv.gameObjects.push(boss);`
This uses an **array** again. The new boss is added to the list of active game objects so it appears in the level.
`const thresholds = [0.75, 0.5, 0.25];`
This array stores boss phase values. Each number represents a health checkpoint for a harder stage of the fight.
`thresholds.forEach((threshold) => { ... })`
This is **iteration**. The game loops through each health checkpoint one by one.
`if (this.bossState.hp <= this.bossState.maxHp * threshold) {`
This combines **conditionals** with a **mathematical expression**. It checks whether the boss has dropped below a percentage of its maximum health.
`summonRushingSharks();`
This is another **function** call. When the health condition is met, the game starts a new boss mechanic by summoning extra enemies.
Overall, this boss section shows how the Aquatic level combines the same CSSE topics from earlier sections, but now uses them to control a full combat encounter instead of a simple dialogue or quest step.
