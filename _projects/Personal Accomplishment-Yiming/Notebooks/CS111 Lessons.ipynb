## Mermaid Boss Explanation

The Mermaid boss is a second combat system layered on top of the Megalodon system. It is a good CS111 example of abstraction, booleans, timed algorithms, object instantiation, and array-based game loops.

```javascript
this.startMermaidBossEncounter = async () => {
  const state = this.mermaidBossState;
  state.active = true;
  state.combatReady = true;
  state.hp = state.maxHp;

  state.volleyShotsRemaining = 0;
  state.nextVolleyReadyAt = Date.now() + 1200;
  state.nextBombAt = Date.now() + 9500;
  state.nextSummonAt = Date.now() + 13500;

  setMermaidBossSpriteSheet(mermaidBossSpriteSrc, { width: 948, height: 948 }, { rows: 6, columns: 6 });
};
```

### Mermaid Code Breakdown

1. Separate state object
- `const state = this.mermaidBossState;` keeps Mermaid logic isolated from Megalodon logic.
- `active` and `combatReady` booleans gate combat updates.

2. Timed scheduling
- `Date.now() + ...` creates future timestamps for volleys, bombs, and summons.
- This is algorithmic control of combat pacing.

3. Sprite metadata abstraction
- `setMermaidBossSpriteSheet(...)` passes source, pixel size, and row/column data.
- This keeps rendering rules data-driven instead of hardcoded.

4. Guardian instantiation and registration

```javascript
const guardianData = {
  orientation: { rows: 6, columns: 6 },
  idle: { row: 0, start: 0, columns: 6 },
  walk: { row: 1, start: 0, columns: 6 },
  attack: { row: 4, start: 0, columns: 5 },
  rangedAttack: { row: 5, start: 0, columns: 5 }
};

const guardian = new Npc(guardianData, this.gameEnv);
this.gameEnv.gameObjects.push(guardian);
```

- `new Npc(...)` is class construction.
- `gameObjects.push(...)` inserts the object into the engine's update/render array.

5. Table-driven animation mapping
- Action-to-row mapping (`idle`, `walk`, `attack`, `rangedAttack`) is cleaner than hardcoding logic per frame.
- Mixed frame counts (6 vs 5) avoid frame bleed across sprite rows.
